```python
import pandas as pd
import os

# 四个正式模型的 results.csv 路径
model_csv = {
    "YOLOv8-S": "/content/drive/MyDrive/fire_smoke_project/runs/yolov8s_640_100e_b16_s42/results.csv",
    "YOLOv9-S": "/content/drive/MyDrive/fire_smoke_project/runs/yolov9s_formal_100e/results.csv",
    "YOLO11-S": "/content/drive/MyDrive/fire_smoke_project/runs/yolo11s_640_100e_b16_s42/results.csv",
    "YOLO26-S": "/content/drive/MyDrive/fire_smoke_project/runs/yolo26s_640_100e_b16_s42/results.csv",
}

# 逐个读取并检查
for model_name, csv_path in model_csv.items():
    print("=" * 80)
    print(model_name)
    print("Path:", csv_path)

    df = pd.read_csv(csv_path)

    print("Epoch 数量:", len(df))
    print("\nCSV 中包含的列：")
    print(df.columns.tolist())

    print("\n最后 3 个 epoch：")
    display(df.tail(3))
```

```python
import pandas as pd
import numpy as np

summary_results = []

for model_name, csv_path in model_csv.items():

    df = pd.read_csv(csv_path)
    df.columns = df.columns.str.strip()

    required_columns = [
        "epoch",
        "metrics/precision(B)",
        "metrics/recall(B)",
        "metrics/mAP50(B)",
        "metrics/mAP50-95(B)"
    ]

    missing = [col for col in required_columns if col not in df.columns]

    if missing:
        print(f"❌ {model_name} 缺少列：{missing}")
        continue

    # 以 mAP50-95 最高的 epoch 作为该模型的最佳性能 epoch
    best_idx = df["metrics/mAP50-95(B)"].idxmax()
    best_row = df.loc[best_idx]

    precision = best_row["metrics/precision(B)"]
    recall = best_row["metrics/recall(B)"]

    if precision + recall > 0:
        f1 = 2 * precision * recall / (precision + recall)
    else:
        f1 = 0

    summary_results.append({
        "Model": model_name,
        "Best Epoch": int(best_row["epoch"]),
        "Precision": precision,
        "Recall": recall,
        "F1-score": f1,
        "mAP@50": best_row["metrics/mAP50(B)"],
        "mAP@50-95": best_row["metrics/mAP50-95(B)"]
    })

summary_df = pd.DataFrame(summary_results)

display_df = summary_df.copy()

for col in ["Precision", "Recall", "F1-score", "mAP@50", "mAP@50-95"]:
    display_df[col] = display_df[col].round(4)

print("=" * 90)
print("FOUR-MODEL BEST PERFORMANCE COMPARISON")
print("=" * 90)

display(display_df)
```

```python
from ultralytics import YOLO
import pandas as pd
import os
import gc
import torch

complexity_results = []

for model_name, weight_path in model_weights.items():

    print("\n" + "=" * 70)
    print(f"Processing: {model_name}")
    print("=" * 70)

    try:
        # 读取模型
        model = YOLO(weight_path)

        # 文件大小
        file_size_mb = os.path.getsize(weight_path) / (1024 ** 2)

        # Fuse 模型，使复杂度统计更接近实际推理结构
        model.fuse(verbose=False)

        # 获取模型信息
        info = model.info(
            detailed=False,
            verbose=True,
            imgsz=640
        )

        layers, parameters, gradients, gflops = info

        complexity_results.append({
            "Model": model_name,
            "Parameters": parameters,
            "Parameters (M)": parameters / 1e6,
            "GFLOPs": gflops,
            "Model Size (MB)": file_size_mb,
            "Layers": layers
        })

        print(f"\n✅ {model_name}")
        print(f"Parameters : {parameters:,}")
        print(f"Parameters : {parameters / 1e6:.2f} M")
        print(f"GFLOPs     : {gflops:.2f}")
        print(f"Model Size : {file_size_mb:.2f} MB")
        print(f"Layers     : {layers}")

        # 清理
        del model
        gc.collect()

        if torch.cuda.is_available():
            torch.cuda.empty_cache()

    except Exception as e:
        print(f"❌ Error loading {model_name}:")
        print(e)


# 转成 DataFrame
complexity_df = pd.DataFrame(complexity_results)

if len(complexity_df) > 0:

    display_complexity = complexity_df.copy()

    display_complexity["Parameters (M)"] = (
        display_complexity["Parameters (M)"].round(2)
    )

    display_complexity["GFLOPs"] = (
        display_complexity["GFLOPs"].round(2)
    )

    display_complexity["Model Size (MB)"] = (
        display_complexity["Model Size (MB)"].round(2)
    )

    print("\n" + "=" * 90)
    print("MODEL COMPLEXITY COMPARISON")
    print("=" * 90)

    display(
        display_complexity[
            [
                "Model",
                "Parameters (M)",
                "GFLOPs",
                "Model Size (MB)",
                "Layers"
            ]
        ]
    )
```

```python
import glob
import os

print("=" * 90)
print("SEARCHING FOR YOLO26 DATA CONFIG")
print("=" * 90)

search_patterns = [
    "/content/drive/MyDrive/**/data_yolo26.yaml",
    "/content/drive/MyDrive/**/*yolo26*.yaml",
    "/content/**/data_yolo26.yaml"
]

found_files = []

for pattern in search_patterns:
    files = glob.glob(pattern, recursive=True)

    for file in files:
        if file not in found_files:
            found_files.append(file)

if len(found_files) == 0:
    print("❌ No YOLO26 data yaml found.")
else:
    print(f"✅ Found {len(found_files)} possible file(s):\n")

    for i, path in enumerate(found_files, 1):
        print(f"{i}. {path}")

        try:
            print("-" * 70)
            with open(path, "r") as f:
                print(f.read())
            print("-" * 70)

        except Exception as e:
            print("Could not read:", e)
```

```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import os

cm_paths = {
    "YOLOv8-S": "/content/drive/MyDrive/fire_smoke_project/runs/yolov8s_640_100e_b16_s42/confusion_matrix_normalized.png",
    "YOLOv9-S": "/content/drive/MyDrive/fire_smoke_project/runs/yolov9s_formal_100e/confusion_matrix_normalized.png",
    "YOLO11-S": "/content/drive/MyDrive/fire_smoke_project/runs/yolo11s_640_100e_b16_s42/confusion_matrix_normalized.png",
    "YOLO26-S": "/content/drive/MyDrive/fire_smoke_project/runs/yolo26s_640_100e_b16_s42/confusion_matrix_normalized.png",
}

fig, axes = plt.subplots(2, 2, figsize=(13, 11))

for ax, (model_name, path) in zip(axes.flat, cm_paths.items()):

    img = mpimg.imread(path)

    ax.imshow(img)
    ax.set_title(model_name, fontsize=13)
    ax.axis("off")

plt.suptitle(
    "Normalized Confusion Matrices of Four YOLO Models",
    fontsize=16
)

plt.tight_layout(rect=[0, 0, 1, 0.96])

save_path = (
    "/content/drive/MyDrive/fire_smoke_project/"
    "model_comparison/08_normalized_confusion_matrices.png"
)

plt.savefig(
    save_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("✅ Figure saved:")
print(save_path)
```
